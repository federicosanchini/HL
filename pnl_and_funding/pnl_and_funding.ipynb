```python
import pandas as pd
import matplotlib.pyplot as plt

# Carica il CSV
df_funding = pd.read_csv("trades_export_crowd.csv")

# Converte la colonna temporale in datetime
df_funding["close_date"] = pd.to_datetime(df_funding["close_date"])

# Ordina per tempo
df_funding = df_funding.sort_values("close_date")

# Funding totale
totale_funding = df_funding["funding"].sum()
print("Funding totale:", totale_funding)

# Funding cumulativo
df_funding["funding_cum"] = df_funding["funding"].cumsum()

# Grafico funding cumulativo nel tempo
plt.figure()
plt.plot(df_funding["close_date"], df_funding["funding_cum"])
plt.xlabel("Time")
plt.ylabel("Funding cumulato (USD)")
plt.title("Funding cumulativo nel tempo")
plt.show()
```
